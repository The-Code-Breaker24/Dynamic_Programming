Complete Bar Plot Function Reference Table:

| `plt.ylabel()`          | Set y-axis label                              | `plt.ylabel('Sales')`                      |

A. Bar Color Parameters (plt.bar() / ax.bar()):
| Parameter    | Purpose                 | Example                          |
| ------------ | ----------------------- | -------------------------------- |
| `color=`     | Main fill color of bars | `plt.bar(x,y,color='red')`       |
| `facecolor=` | Interior color of bars  | `plt.bar(x,y,facecolor='blue')`  |
| `edgecolor=` | Border color of bars    | `plt.bar(x,y,edgecolor='black')` |
| `alpha=`     | Transparency            | `plt.bar(x,y,alpha=0.5)`         |
| `hatch=`     | Pattern inside bars     | `plt.bar(x,y,hatch='//')`        |
| `linewidth=` | Border thickness        | `plt.bar(x,y,linewidth=3)`       |
